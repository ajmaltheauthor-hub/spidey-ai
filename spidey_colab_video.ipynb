```python
# Step 1: Mount your Google Drive
from google.colab import drive
drive.mount('/content/drive')

import os
VIDEO_OUT = '/content/drive/MyDrive/Spidey-AI/videos'
os.makedirs(VIDEO_OUT, exist_ok=True)
print('✅ Drive Mounted! Videos will save to:', VIDEO_OUT)
```

```python
# Step 2: Install LTX-Video / Diffusers for Free Cloud GPU Video Generation
!pip install -q diffusers transformers accelerate imageio[ffmpeg] sentencepiece
import torch
print('GPU Available:', torch.cuda.get_device_name(0) if torch.cuda.is_available() else 'Switch Runtime -> Change runtime type -> T4 GPU!')
```

```python
# Step 3: Generate AI Video & Save Straight to G:\My Drive\Spidey-AI\videos
import time, torch
from diffusers import LTXPipeline
from diffusers.utils import export_to_video

pipe = LTXPipeline.from_pretrained('Lightricks/LTX-Video', torch_dtype=torch.bfloat16)
pipe.to('cuda')

prompt = 'Cinematic neon cyberpunk street in heavy rain, Spider-style hero swinging past glowing billboards, 4k photorealistic'
video = pipe(
    prompt=prompt,
    negative_prompt='worst quality, blurry, jittery, distorted',
    width=768,
    height=512,
    num_frames=97,
    num_inference_steps=35
).frames[0]

out_path = f'{VIDEO_OUT}/spidey_colab_{int(time.time())}.mp4'
export_to_video(video, out_path, fps=24)
print('🎉 Video saved straight to Google Drive:', out_path)
```
